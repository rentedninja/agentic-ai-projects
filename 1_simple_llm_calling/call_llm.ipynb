```python
from dotenv import load_dotenv
from langchain_google_genai import ChatGoogleGenerativeAI


load_dotenv()
```

```python
llm = ChatGoogleGenerativeAI(model="gemma-4-31b-it")

response = llm.invoke("Write a poem about the ocean.")
print(response.text)
```

```python
response = llm.invoke([
    ["system", "You are a helpful assistant that answers in one line"],
    ["human","how many moons does Jupiter have?"]
])
print(response.text)
```

```python
prompt = "Write a single tagline for a quirky coffee shop. Only output the tagline,nothing else."

llm = ChatGoogleGenerativeAI(model="gemini-3.8-flash", temperature=0.5)
response = llm.invoke(prompt)
print(response.text)
```
